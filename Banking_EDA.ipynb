```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\NAJISH\Downloads\Banking (1).csv")
df.head(4)
```

```python
# Generate the descriptive statistics of data
df.describe()
```

```python
df['Age'].min()
```

```python
bins=[0,100000,300000,float('inf')]
labels=['low','medium','high']
```

```python
df['Income Band']=pd.cut(df['Estimated Income'],bins=bins,labels=labels,right=False)
```

```python
df.groupby('Income Band').agg({'Estimated Income':'sum'})
```

```python
df['Income Band'].value_counts().plot(kind='bar',color='g')
```

```python
df.groupby('Nationality').agg({'Estimated Income':'sum'}).plot(kind='barh')
```

```python
df.groupby('Amount of Credit Cards').agg({'Amount of Credit Cards':'count'})
```

# Univariate Analysis

```python
# Examine the distribution of unique categories in categorical columns.
categorical_cols=df[['BRId','GenderId','IAId','Nationality','Amount of Credit Cards','Occupation','Fee Structure','Loyalty Classification','Income Band','Risk Weighting']].columns

for col in categorical_cols:
    print(f"value of counts for '{col}':")
    display(df[col].value_counts())
```

```python
for i, predictor in enumerate (df[['BRId','GenderId','IAId','Nationality','Amount of Credit Cards','Occupation','Fee Structure','Loyalty Classification','Income Band','Risk Weighting']]):
    plt.figure(i)
    sns.countplot(data=df,x=predictor)
```

# Bivariate Analysis

```python
for i, predictor in enumerate (df[['BRId','GenderId','IAId','Nationality','Amount of Credit Cards','Occupation','Fee Structure','Loyalty Classification','Income Band','Risk Weighting']]):
    plt.figure(i)
    sns.countplot(data=df,x=predictor,hue='Nationality')
```

```python
# Histogram of value counts for different occupations

for col in categorical_cols:
    if col=='Occupation':
        continue
    plt.figure(figsize=(8,4))
    sns.histplot(df[col])
    plt.title("Histogram Of Occupation Counts")
    plt.xlabel(col)
    plt.ylabel("Counts")
    plt.show()
    
```

# Numerical Analysis

```python
numerical_cols=['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Deposits','Saving Accounts','Checking Accounts']
plt.figure(figsize=(30,15))
for i , col in enumerate(numerical_cols):
    plt.subplot(6,3,i+1)
    sns.histplot(df[col],kde=True)
    plt.title(col)
plt.show()
```

```python
numerical_cols=['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Deposits','Saving Accounts','Checking Accounts']
correlation_matrix=df[numerical_cols].corr()
plt.figure(figsize=(8,8))
sns.heatmap(correlation_matrix,annot=True,cmap='crest',fmt='.2f')
plt.title("Correlation Matrix")
plt.show()
```
